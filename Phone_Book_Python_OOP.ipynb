```python
# NOTE: class, method and variable names are kept in Italian, as in the original project.
# The menu shown to the user is generated dynamically from the method names (see Rubrica.functionlaties),
# so renaming them would change the program. Glossary:
#   Contatto = Contact, Rubrica = Phone book, nome = first name, cognome = last name, numero = number, prefisso = country code,
#   cerca = search, ricerca_contatto = search contact, aggiungi_contatto = add contact, visualizza_contatti = view contacts,
#   modifica_contatto = edit contact, elimina_contatto = delete contact, esci = exit
class Contatto:


  """
  Definition of the Contatto (Contact) class, which holds the contact details: First name, Last name, Number, Country code

  """

#############################################################################################################################
  #the default country code is the Italian one, 39; the last name is optional (empty by default)
  def __init__(self, nome, numero,cognome = '',prefisso = 39):

    """
    Initialize the attributes of the Contact.
    -nome: the first name of the contact (string)
    -cognome: the last name of the contact (string)
    -numero: the phone number (number)
    -prefisso: the country calling code (number)

    A contact is uniquely identified by a First name, Last name, Country code and Number

    """

    self._nome = nome
    self._cognome = cognome

    #number and country code must be numeric
    try:
      self._numero = int(numero)
      self._prefisso = int(prefisso)
    except ValueError:
      # Block the creation of the object
      raise ValueError("Error: The number and the country code must be numeric values.")

#############################################################################################################################
  def __eq__(self,other_contact):

    """
    Checks whether two contacts are equal by comparing all their attributes.
    """
    #check whether other_contact is of type Contatto
    if not isinstance(other_contact, Contatto):
      return False
    #return True if every element of the two contacts is equal

    return self._nome == other_contact._nome and self._numero == other_contact._numero and self._cognome == other_contact._cognome and self._prefisso == other_contact._prefisso

#############################################################################################################################
  def __lt__(self, other_contact):

        """
        Method used to sort the contact list by last name/first name
        """

        # Check that we are comparing with another Contatto
        if not isinstance(other_contact, Contatto):
            return NotImplemented

        # Sort by last name first. If the last names are equal, sort by first name
        return (self._cognome, self._nome) < (other_contact._cognome, other_contact._nome)



#############################################################################################################################


  def input_valid_contact():
    """
        Method to read a valid Contact from user input
    """

    name = input("Enter the first name: ")
    surname = input("Enter the last name: ")
    dictionary_number_prefix = {"number": None, "country code": None}


    for chiave in dictionary_number_prefix:
      input_contact= False
      while not input_contact:
        try:
          dictionary_number_prefix[chiave] = int(input(f"Enter the {chiave}: "))
          input_contact = True
        except ValueError:
          print("Enter a valid number")


    return Contatto(name, dictionary_number_prefix["number"], surname, dictionary_number_prefix["country code"])
```

```python
class Rubrica:

  """
  Definition of the methods that make up a phone book
  The main rule of this phone book is that there cannot be two absolutely identical contacts
  """
#############################################################################################################################
  def __init__(self, rubrica = None): #I assign None as the default value because otherwise Rubrica instances created without passing an argument would share the same list in memory
    if rubrica is None:
      self._rubrica = []
    else:
      self._rubrica = rubrica






#############################################################################################################################
  def cerca_contatto(self,contatto):

    """
    Checks whether the contact is part of the phone book or not
    """

    is_contact_in_list = False


    #check whether the argument is a Contatto
    if isinstance(contatto, Contatto):

      #loop to check whether the contact is present, distinguishing whether only First/Last name or number or both are present
      #exit the loop if the contact is present

      if self._rubrica == []:
        print("The phone book has no saved contacts")
      else:
        for contact in self._rubrica:
          if contact == contatto:
            #print(f"The contact {contatto._nome} {contatto._cognome} with number + {contatto._prefisso} {contatto._numero} is already saved in the phone book")
            is_contact_in_list = True
            return is_contact_in_list
            break
          else:
            continue

    else:
      print("It is not a Contatto")
      return is_contact_in_list


#############################################################################################################################
  def cerca_name(self, name):

    name_list = []
    print(f"Results for the search of {name}:\n")
    for contact in self._rubrica:
      if contact._nome == name or contact._cognome == name:
        print(f"{contact._nome} {contact._cognome}: +{contact._prefisso} {contact._numero} \n")
        name_list.append(contact)

    if name_list == []:
      print("No contact found by the search")
    #return name_list




#############################################################################################################################
  def cerca_number(self, prefix,number):

    number_list = []
    print(f"Results for the search of +{prefix} {number}:\n")
    for contact in self._rubrica:
      if contact._prefisso == prefix and contact._numero == number:
        print(f"{contact._nome} {contact._cognome}: +{contact._prefisso} {contact._numero} \n")
        number_list.append(contact)

    if number_list == []:
      print("No contact found by the search")
    #return number_list


#############################################################################################################################
  def ricerca_contatto(self):

    """
    Method to search for a contact in the phone book, either by first/last name or by number/country code
    """

    is_valid_input = False

    while not is_valid_input:
      try:
        is_researching = True
        while is_researching:
          ricerca = int(input("Type 1 to search by name, 2 to search by number, 3 to stop searching"))
          if ricerca == 1:
            #is_valid_number = True
            name = input("Enter the first or last name of the contact to search for: ")
            self.cerca_name(name)
          elif ricerca == 2:

            is_valid_prefix = False
            is_valid_number = False

            while not(is_valid_prefix) or not(is_valid_number):
              try:
                prefix = int(input("Enter the country code of the contact to search for: "))
                is_valid_prefix = True

                number = int(input("Enter the number of the contact to search for: "))
                is_valid_number = True
              except ValueError:
                print("Enter a valid phone number (country code + number)")
            self.cerca_number(prefix, number)
          elif ricerca == 3:
            is_researching = False
            print("Search stopped")

          else:
            print("Enter a valid number: 1, 2 or 3 \n")
        is_valid_input = True
      except ValueError:
        print("Enter a valid number: 1, 2 or 3 \n")


#############################################################################################################################
  def aggiungi_contatto(self):

    """
    Method to add a contact from user input, first checking whether the contact is already in the phone book
    """

    #read the contact I want to add
    contatto = Contatto.input_valid_contact()


    #check that the contact is not already part of the list; if it is not, add it
    if not self.cerca_contatto(contatto):

      self._rubrica.append(contatto)
      print(f"The contact {contatto._nome} {contatto._cognome} has been added \n")


    else:
      print("Contact already saved in the phone book \n")



#############################################################################################################################
  def aggiungi_contatto_da_file(self,contatto):

    """
    Method to add a contact from a File
    """

    if not self.cerca_contatto(contatto):
    #check that the contact is not already part of the list; if it is not, add it
      self._rubrica.append(contatto)
      print(f"The contact {contatto._nome} {contatto._cognome} has been added")

    else:
      print("Contact already saved in the phone book \n")




#############################################################################################################################
  def visualizza_contatti(self):
    """
    Method to display the contact list in alphabetical order
    """
    self._rubrica.sort() #sort the phone book in alphabetical order

    print("CONTACTS\n\n")

    for contatto in self._rubrica:

      print(f"{contatto._nome} {contatto._cognome}: +{contatto._prefisso} {contatto._numero} \n")





#############################################################################################################################
  def modifica_contatto(self):
    """
    Method to edit a contact of a phone book, after checking that the contact is in the phone book
    """

    print("Which contact do you want to edit?")

    #read the contact I want to edit
    contatto1 = Contatto.input_valid_contact()



    if self.cerca_contatto(contatto1): #if it exists, edit it

      is_duplicated = True #flag for duplicated records
      exit = 'N' #variable that allows exiting the following loop based on the user's decision
      contatore = 1
      print(f"EDIT THE CONTACT [{contatto1._nome} {contatto1._cognome}: +{contatto1._prefisso} {contatto1._numero}] \n")

      while is_duplicated: #loop to repeat the input in case of a duplicate entry

        if contatore >= 2: #do not allow exiting the loop until the second attempt
          exit = input("Type Y if you want to exit EDIT mode. To continue, type any other character: ")#exit is updated according to the user's choice

        if exit == 'Y': #exit the loop
          break
        else:

          contatto2 = Contatto.input_valid_contact()

          if contatto2 in self._rubrica:
            print(f"The contact {contatto2._nome} {contatto2._cognome}: +{contatto2._prefisso} {contatto2._numero} is already saved in the phone book. This edit is not possible\n") #duplicate contacts are not allowed in the phone book
            contatore+=1
          else:

            i= self._rubrica.index(contatto1)
            self._rubrica[i] = contatto2
            is_duplicated = False

            print(f"The contact [{contatto1._nome} {contatto1._cognome}: +{contatto1._prefisso} {contatto1._numero}] has been changed to [{contatto2._nome} {contatto2._cognome}: +{contatto2._prefisso} {contatto2._numero}]\n")


    else:
      print(f" The contact {contatto1._nome} {contatto1._cognome}: +{contatto1._prefisso} {contatto1._numero} is not in the phone book\n")





#############################################################################################################################
  def elimina_contatto(self):

    """
    Method to remove a contact from a phone book, after checking that the contact is in the phone book
    """
    contatto = Contatto.input_valid_contact()

    if self.cerca_contatto(contatto):
      self._rubrica.remove(contatto)
      print(f" The contact [{contatto._nome} {contatto._cognome}: +{contatto._prefisso} {contatto._numero}] has been deleted from the phone book\n")


    else:
      print(f" The contact {contatto._nome} {contatto._cognome}: +{contatto._prefisso} {contatto._numero} is not in the phone book\n")






#############################################################################################################################
  def load_file_in_rubr(self,file_path):

    """
    Method to load the phone book from a file
    """

    if os.path.exists(file_path):
      try:
        with open(file_path,'r') as file:
            list_rubrica = json.load(file) #load the phone book from the file into a temporary list
            print("Loading contacts from File \n")

            for contact in list_rubrica:

              contatto = Contatto(contact["nome"],contact["numero"],contact["cognome"], contact["prefisso"]) #define the Contatto
              self.aggiungi_contatto_da_file(contatto) #add the defined contact

            print("Contacts loaded from File \n")

        return self._rubrica
      except json.JSONDecodeError:
        print(f"Error: The file '{file_path}' is not a valid JSON file.")




    else:
      print(f"The file '{file_path}' was not found.")








#############################################################################################################################

  def convert_rubr_dict(self):

    """
    Method to convert the phone book list into a list of dictionaries
    """

    #automatically retrieve the attributes that define a contact
    nome_attributi = inspect.signature(Contatto.__init__)
    argomenti = [attributo for attributo in nome_attributi.parameters if attributo != 'self']

    lista_rubrica = []

    #build the list of contacts; each contact is a dictionary made of the arguments defined above
    for contact in self._rubrica:
      dizionario_contatto = {}
      for attributes in argomenti:

        attr = '_'+attributes
        dizionario_contatto[attributes] = getattr(contact, attr) #dynamically extract the attribute names to assign the corresponding value of the specific contact

      lista_rubrica.append(dizionario_contatto)


    return lista_rubrica


#############################################################################################################################

  def load_rubr_in_file(self,file_path):

    """
    Method to write the list of dictionaries to File, overwriting it
    """
    #transform the Rubrica into a list of contact dictionaries
    lista_rubrica = self.convert_rubr_dict()


    with open(file_path,'w') as file:
      json.dump(lista_rubrica, file, indent = 4)

    print("Contacts saved successfully")




#############################################################################################################################
  def functionlaties(self):
    """
    Method to list all the features of the phone book (ricerca_contatto, aggiungi_contatto, visualizza_contatto,...)
    """
    lista_metodi = []
    metodi_esclusi = ["cerca_name","cerca_number","cerca_contatto","_rubrica","functionlaties","aggiungi_contatto_da_file","load_file_in_rubr","convert_rubr_dict","load_rubr_in_file"] #these methods are used internally, they are not user features
    for nome_metodo in dir(self):

      if nome_metodo not in metodi_esclusi and  not nome_metodo.startswith("__"):
        dict_function = {nome_metodo.replace('_',' ').upper(): nome_metodo}
        lista_metodi.append(dict_function)

    return lista_metodi


#############################################################################################################################
  def esci(self):
    """
    Method to exit the phone book
    """

    return None
```

```python
import json
import os
import inspect



"""
  Main function that manages the program flow.
  1. Load the phone book from file;
  2. Phone book features;
  3. Save the phone book to file
"""
#PHONE BOOK INSTANTIATION
rubr = Rubrica()

################################################################################
#LOAD THE PHONE BOOK FROM FILE
file_path = "/content/drive/MyDrive/Repository Master/Progetto Python/rubrica.json"

rubr.load_file_in_rubr(file_path)


################################################################################
#PROGRAM BODY (ADD, EDIT, SEARCH, DELETE, VIEW)


accesso_rubrica = input("Do you want to access your Phone book? Y/n: ")

print("Phone book accessed \n") if accesso_rubrica == 'Y' else print("Phone book not accessed")

funzionalita = rubr.functionlaties()


while accesso_rubrica == 'Y':


  print("FEATURES \n")
  for i, dizionario_funzione in enumerate(funzionalita):
    for chiave in dizionario_funzione:
        print(f"{i+1}) {chiave}")

  try:
    i = int(input("What do you want to do? (choose the number) \n"))
    assert(i >=1 and i <= len(funzionalita))
    dizionario_selezionato = funzionalita[i-1]
    funzione_selezionata = list(dizionario_selezionato.values())[0]
    metodo_classe = getattr(rubr, funzione_selezionata)

    if funzione_selezionata == 'esci':
      print("Closing the Phone book")
      break
    else:
      metodo_classe()

  except ValueError:
    print("Enter a numeric value \n")
  except AssertionError:
    print("Choose one of the available options \n")

################################################################################
#SAVE THE PHONE BOOK


rubr.load_rubr_in_file(file_path)
```
