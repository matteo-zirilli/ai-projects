```python
class Contatto:


  """
  Definizione della classe Contatto dove sono presenti gli elementi di contatto: Nome, Cognome, Numero, Prefisso

  """

#############################################################################################################################
  #il prefisso di default è quello italiano 39, cognome non di default non dichiarato
  def __init__(self, nome, numero,cognome = '',prefisso = 39):

    """
    Inizializzo gli attributi del Contatto.
    -nome: il nome del contatto (stringa)
    -cognome: il cognome del contatto (stringa)
    -numero: il numero telefonico (numero)
    -prefisso: prefisso telefonico (numero)

    Un contatto è univocamente definito o da un Nome, Cognome, Prefisso e Numero

    """

    self._nome = nome
    self._cognome = cognome

    #numero e prefisso devono essere numerici
    try:
      self._numero = int(numero)
      self._prefisso = int(prefisso)
    except ValueError:
      # Blocco la creazione dell'oggetto
      raise ValueError("Errore: Il numero e il prefisso devono essere valori numerici.")

#############################################################################################################################
  def __eq__(self,other_contact):

    """
    Verifica l'uguaglianza tra due contatti confrontando tutti i loro attributi.
    """
    #verifico se other_contact è tipo Contatto
    if not isinstance(other_contact, Contatto):
      return False
    #ritorno true se ogni elemento dei due contatti è uguale

    return self._nome == other_contact._nome and self._numero == other_contact._numero and self._cognome == other_contact._cognome and self._prefisso == other_contact._prefisso

#############################################################################################################################
  def __lt__(self, other_contact):

        """
        Metodo per ordinare la lista dei contatti in base a cognome/nome
        """

        # Controllo che stiamo confrontando con un altro Contatto
        if not isinstance(other_contact, Contatto):
            return NotImplemented

        # Ordina prima per cognome. Se i cognomi sono uguali, ordina per nome
        return (self._cognome, self._nome) < (other_contact._cognome, other_contact._nome)



#############################################################################################################################


  def input_valid_contact():
    """
        Metodo per assegnare in input un Contatto Valido
    """

    name = input("Inserisci il nome: ")
    surname = input("Inserisci il cognome: ")
    dictionary_number_prefix = {"numero": None, "prefisso": None}


    for chiave in dictionary_number_prefix:
      input_contact= False
      while not input_contact:
        try:
          dictionary_number_prefix[chiave] = int(input(f"Inserisci il {chiave}: "))
          input_contact = True
        except ValueError:
          print("Inserisci un numero valido")


    return Contatto(name, dictionary_number_prefix["numero"], surname, dictionary_number_prefix["prefisso"])
```

```python
class Rubrica:

  """
  Definzione dei metodi che compongono una rubrica
  La principale regola di questa rubrica è che non ci possono essere due contatti assolutamente identici
  """
#############################################################################################################################
  def __init__(self, rubrica = None): #come valore di default assegno None perché altrimenti le istanze Rubrica create senza passare un argomento condivideranno la stessa lista in memoria
    if rubrica is None:
      self._rubrica = []
    else:
      self._rubrica = rubrica






#############################################################################################################################
  def cerca_contatto(self,contatto):

    """
    Verifica se il contatto va parte della rubrica o meno
    """

    is_contact_in_list = False


    #verifica se l'argomento è un Contatto
    if isinstance(contatto, Contatto):

      #ciclo per verificare che il contatto sia presente, distinzione se è presente solo Nome/Cognome o numero o entrambi
      #esco dal ciclo se il contatto è presente

      if self._rubrica == []:
        print("La rubrica non ha nessun contatto registrato")
      else:
        for contact in self._rubrica:
          if contact == contatto:
            #print(f"Il contatto {contatto._nome} {contatto._cognome} con numero + {contatto._prefisso} {contatto._numero} è già registrato in rubrica")
            is_contact_in_list = True
            return is_contact_in_list
            break
          else:
            continue

    else:
      print("Non è un Contatto")
      return is_contact_in_list


#############################################################################################################################
  def cerca_name(self, name):

    name_list = []
    print(f"Risultato per la ricerca di {name}:\n")
    for contact in self._rubrica:
      if contact._nome == name or contact._cognome == name:
        print(f"{contact._nome} {contact._cognome}: +{contact._prefisso} {contact._numero} \n")
        name_list.append(contact)

    if name_list == []:
      print("Nessun contatto trovato dalla ricerca effettuata")
    #return name_list




#############################################################################################################################
  def cerca_number(self, prefix,number):

    number_list = []
    print(f"Risultato per la ricerca di +{prefix} {number}:\n")
    for contact in self._rubrica:
      if contact._prefisso == prefix and contact._numero == number:
        print(f"{contact._nome} {contact._cognome}: +{contact._prefisso} {contact._numero} \n")
        number_list.append(contact)

    if number_list == []:
      print("Nessun contatto trovato dalla ricerca effettuata")
    #return number_list


#############################################################################################################################
  def ricerca_contatto(self):

    """
    Metodo per eseguire la ricerca di un contatto nella rubrica, o per nome/cognome o per numero/prefisso
    """

    is_valid_input = False

    while not is_valid_input:
      try:
        is_researching = True
        while is_researching:
          ricerca = int(input("Digita 1 per eseguire ricerca per nome, 2 per numero,3 per interrompere la ricerca"))
          if ricerca == 1:
            #is_valid_number = True
            name = input("Inserisci il nome o il cognome del contatto da ricercare: ")
            self.cerca_name(name)
          elif ricerca == 2:

            is_valid_prefix = False
            is_valid_number = False

            while not(is_valid_prefix) or not(is_valid_number):
              try:
                prefix = int(input("Inserisci il prefisso del contatto da ricercare: "))
                is_valid_prefix = True

                number = int(input("Inserisci il numero del contatto da ricercare: "))
                is_valid_number = True
              except ValueError:
                print("Inserisci un numero telefonico valido (prefisso + numero)")
            self.cerca_number(prefix, number)
          elif ricerca == 3:
            is_researching = False
            print("Ricerca interrotta")

          else:
            print("Inserisci un numero valido: 1, 2 o 3 \n")
        is_valid_input = True
      except ValueError:
        print("Inserisci un numero valido: 1, 2 o 3 \n")


#############################################################################################################################
  def aggiungi_contatto(self):

    """
    Metodo per aggiungere un contatto in input e verifico prima se il contatto è presente in Rubrica
    """

    #prendo in input il contatto che voglio aggiungere
    contatto = Contatto.input_valid_contact()


    #verifico che il contatto non faccia già parte della lista, se non fa parte lo aggiungo
    if not self.cerca_contatto(contatto):

      self._rubrica.append(contatto)
      print(f"Il contatto {contatto._nome} {contatto._cognome} è stato aggiunto \n")


    else:
      print("Contatto già registrato in Rubrica \n")



#############################################################################################################################
  def aggiungi_contatto_da_file(self,contatto):

    """
    Metodo per aggiungere un contatto da File
    """

    if not self.cerca_contatto(contatto):
    #verifico che il contatto non faccia già parte della lista, se non fa parte lo aggiungo
      self._rubrica.append(contatto)
      print(f"Il contatto {contatto._nome} {contatto._cognome} è stato aggiunto")

    else:
      print("Contatto già registrato in Rubrica \n")




#############################################################################################################################
  def visualizza_contatti(self):
    """
    Metodo per visualizzare la lista contatti in ordine alfabetico
    """
    self._rubrica.sort() #ordino la rubrica in ordine alfabetico

    print("CONTATTI\n\n")

    for contatto in self._rubrica:

      print(f"{contatto._nome} {contatto._cognome}: +{contatto._prefisso} {contatto._numero} \n")





#############################################################################################################################
  def modifica_contatto(self):
    """
    Metodo per modificare un contatto da una rubrica, previa verifica che il contatto sia in Rubrica
    """

    print("Quale contatto vuoi modificare?")

    #prendo in input il contatto che voglio modificare
    contatto1 = Contatto.input_valid_contact()



    if self.cerca_contatto(contatto1): #se esiste lo modifico

      is_duplicated = True #controllore per record duplicati
      exit = 'N' #variabile che da la possibilità di uscire dal ciclo seguente in base alla decisione dell'utente
      contatore = 1
      print(f"MODIFICA IL CONTATTO [{contatto1._nome} {contatto1._cognome}: +{contatto1._prefisso} {contatto1._numero}] \n")

      while is_duplicated: #ciclo per reiterare l'input in caso di inserimento duplicato

        if contatore >= 2: #non permetto di uscire da ciclo a meno che non si arriva alla seconda
          exit = input("Digita Y se vuoi uscire dalla modalità MODIFICA. Se vuoi continuare digita qualsiasi altro carattere: ")#exit si aggiorna in base alla volontà dell'utente

        if exit == 'Y': #esco dal ciclo
          break
        else:

          contatto2 = Contatto.input_valid_contact()

          if contatto2 in self._rubrica:
            print(f"Il contatto {contatto2._nome} {contatto2._cognome}: +{contatto2._prefisso} {contatto2._numero} è registrato in rubrica. Questa modifica non è possibile\n") #contatti duplicati non ammessi in rubrica
            contatore+=1
          else:

            i= self._rubrica.index(contatto1)
            self._rubrica[i] = contatto2
            is_duplicated = False

            print(f"Il contatto [{contatto1._nome} {contatto1._cognome}: +{contatto1._prefisso} {contatto1._numero}] è stato modificato in [{contatto2._nome} {contatto2._cognome}: +{contatto2._prefisso} {contatto2._numero}]\n")


    else:
      print(f" Il contatto {contatto1._nome} {contatto1._cognome}: +{contatto1._prefisso} {contatto1._numero} non si trova in rubrica\n")





#############################################################################################################################
  def elimina_contatto(self):

    """
    Metodo per rimuovere un contatto da una rubrica, previa verifica che il contatto sia in Rubrica
    """
    contatto = Contatto.input_valid_contact()

    if self.cerca_contatto(contatto):
      self._rubrica.remove(contatto)
      print(f" Il contatto [{contatto._nome} {contatto._cognome}: +{contatto._prefisso} {contatto._numero}] è stato eliminato dalla rubrica\n")


    else:
      print(f" Il contatto {contatto._nome} {contatto._cognome}: +{contatto._prefisso} {contatto._numero} non si trova in rubrica\n")






#############################################################################################################################
  def load_file_in_rubr(self,file_path):

    """
    Metodo per caricamento da file in rubrica
    """

    if os.path.exists(file_path):
      try:
        with open(file_path,'r') as file:
            list_rubrica = json.load(file) #carico la rubrica da file in una lista di appoggio
            print("Inizio caricamento contatti da File \n")

            for contact in list_rubrica:

              contatto = Contatto(contact["nome"],contact["numero"],contact["cognome"], contact["prefisso"]) #definisco il Contatto
              self.aggiungi_contatto_da_file(contatto) #aggiungo il contatto definito

            print("Caricamento contatti da File completato \n")

        return self._rubrica
      except json.JSONDecodeError:
        print(f"Errore: Il file '{file_path}' non è un file JSON valido.")




    else:
      print(f"Il file '{file_path}' non è stato trovato.")








#############################################################################################################################

  def convert_rubr_dict(self):

    """
    Metodo per convertire la lista rubrica in una lista di dizionari
    """

    #recupero in maniera automatica gli attributi che definiscono una contatto
    nome_attributi = inspect.signature(Contatto.__init__)
    argomenti = [attributo for attributo in nome_attributi.parameters if attributo != 'self']

    lista_rubrica = []

    #costruisco la lista dei contatti, ogni contatto è un dizionario composto dagli argomenti sopra definiti
    for contact in self._rubrica:
      dizionario_contatto = {}
      for attributes in argomenti:

        attr = '_'+attributes
        dizionario_contatto[attributes] = getattr(contact, attr) #estraggo dinamicamente i nome degli attributi per assegnare il valore corrispettivo dello specifico contatto

      lista_rubrica.append(dizionario_contatto)


    return lista_rubrica


#############################################################################################################################

  def load_rubr_in_file(self,file_path):

    """
    Metodo per caricare la lista di dizionari su File in sovrascrittura
    """
    #trasformo la Rubrica in una lista di dizionari contatto
    lista_rubrica = self.convert_rubr_dict()


    with open(file_path,'w') as file:
      json.dump(lista_rubrica, file, indent = 4)

    print("Contatti salvati con successo")




#############################################################################################################################
  def functionlaties(self):
    """
    Metodo per elencare tutte le funzionalità della rubrica (ricerca_contatto, aggiungi_contatto, visualizza_contatto,...)
    """
    lista_metodi = []
    metodi_esclusi = ["cerca_name","cerca_number","cerca_contatto","_rubrica","functionlaties","aggiungi_contatto_da_file","load_file_in_rubr","convert_rubr_dict","load_rubr_in_file"] #tali metodi servono internamente, non sono le funzionalità utente
    for nome_metodo in dir(self):

      if nome_metodo not in metodi_esclusi and  not nome_metodo.startswith("__"):
        dict_function = {nome_metodo.replace('_',' ').upper(): nome_metodo}
        lista_metodi.append(dict_function)

    return lista_metodi


#############################################################################################################################
  def esci(self):
    """
    Metodo per uscire dalla rubrica
    """

    return None
```
